```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

dataset["CATEGORY_encoded"] = encoder.fit_transform(dataset["CATEGORY"])
dataset["PURPOSE_encoded"] = encoder.fit_transform(dataset["PURPOSE"])
```

```python
plt.figure(figsize=(7, 4))
sns.histplot(dataset[dataset["MILES"] < 40]["MILES"], kde=True)
plt.title("Distribution of Ride Miles (Less than 40 miles)")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.boxplot(y=dataset[dataset["MILES"] < 100]["MILES"])
plt.title("Miles Distribution (Less than 100 miles)")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.boxplot(y=dataset["MILES"])
plt.title("Distribution of Ride Miles")
plt.show()
```

```python
dataset["MILES"].describe()
```

```python
plt.figure(figsize=(7, 4))
sns.countplot(x="day", data=dataset,
              order=["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])
plt.title("Rides by Day of Week")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x="day-night", data=dataset)
plt.title("Rides by Time of Day")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(x="PURPOSE", data=dataset)
plt.xticks(rotation=45)
plt.title("Purpose of Uber Rides")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x="CATEGORY", data=dataset)
plt.title("Ride Category Distribution")
plt.show()
```

```python
dataset["day"] = dataset["START_DATE"].dt.weekday
```

```python
dataset["day-night"] = pd.cut(
    dataset["hour"],
    bins=[0, 10, 16, 20, 24],
    labels=["Morning", "Afternoon", "Evening", "Night"],
    right=False
)
```

```python
dataset["hour"] = dataset["START_DATE"].dt.hour
```
